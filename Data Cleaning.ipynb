```python
#concatenate the 12 datasets

year_data = pd.concat([df0425, df0525, df0625, df0725, df0825, df0925, df1025, df1125, df1225, df0126, df0226, df0326], axis = 0) 
```

BEGIN CLEANING 

```python
#convert start and end times to datetime 
year_data['started_at']=pd.to_datetime(year_data['started_at']) 
year_data['ended_at']=pd.to_datetime(year_data['ended_at'])

#create new column ride length that is the difference between end and start time 
year_data ['ride_length'] =(year_data.ended_at)-(year_data.started_at)
print(year_data['ride_length'].dtype) #check that data type is timedelta 
```

```python
#make sure ride length is timedelta 
year_data['ride_length']=pd.to_timedelta(year_data['ride_length'], errors='coerce') 

#ride_length converted to mins  
year_data['ride_length']=(year_data['ride_length'].dt.total_seconds() / 60).round(2) 
```

```python
#create a new column with day of week of each trip 
year_data['day_of_week'] = year_data['started_at'].dt.day_name() 
```

```python
#check data nulls
year_data.isnull().sum()
```

```python
#fill nulls with NO DATA 
year_data=year_data.fillna('NO DATA')
```

```python
#filter for trips longer than 3 hours
long_trip=year_data[year_data.ride_length>180]
long_trip
```
