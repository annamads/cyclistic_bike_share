DAY OF WEEK ANALYSIS 

```python
#Sort days of week in order 
days_order = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
year_data['day_of_week'] = pd.Categorical(year_data['day_of_week'],
categories=days_order, ordered=True)
```

```python
#analyze ride count per day of week per member type 
year_data.groupby(['member_type', 'day_of_week'], observed=True).size().unstack()
```

```python
#analyze avg ride length per day of week per member type 

year_data.groupby(['member_type', 'day_of_week'],observed=True)['ride_length_minutes'].mean().unstack().round(2)
```

```python
#analyze median ride length per day of week per member type 

year_data.groupby(['member_type', 'day_of_week'],observed=True)['ride_length_minutes'].median().unstack().round(2)
```

```python
#create filtered data frame for only weekdays 

weekday=(year_data.day_of_week=="Monday") | (year_data.day_of_week =="Tuesday") | (year_data.day_of_week=='Wednesday') | (year_data.day_of_week=='Thursday') | (year_data.day_of_week=='Friday')

weekday = year_data[weekday]

#find ride count per member type on weekdays 
weekday[['member_type']].value_counts()
```

```python
#create filtered data frame for only weekends  


weekend=(year_data.day_of_week=="Saturday") | (year_data.day_of_week =="Sunday") 

weekend = year_data[weekend]

#find ride count per member type on weekends  
weekend[['member_type']].value_counts()
```

```python
#calculate proportion of rides that happened on weekdays vs weekends for each member type 
casual_weekday=round((1220243/1947077)*100,2)
member_weekday=round((2723336/3547953)*100,2)
casual_weekend=round((726834/1947077)*100,2)
member_weekend=round((824617/3547953)*100,2)

print (casual_weekday)
print (casual_weekend)
print (member_weekday) 
print (member_weekend)
```

```python
#find average ride length for each member type on weekdays
weekday.groupby(['member_type'],observed=True)['ride_length_minutes'].mean().round(2)
```

```python
#find average ride length for each member type on weekends 

weekend.groupby(['member_type'],observed=True)['ride_length_minutes'].mean().round(2)
```

MONTH OF YEAR ANALYSIS

```python
#Sort months in order 

months_order = ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]
year_data['ride_month'] = pd.Categorical(year_data['ride_month'],
categories=months_order, ordered=True)
```

```python
#find ride count for each member type per month 

year_data.groupby(['member_type', 'ride_month'],observed=True)['ride_length_minutes'].size().unstack()
```

```python
#find average ride length for each member type per month 

year_data.groupby(['member_type', 'ride_month'],observed=True)['ride_length_minutes'].mean().unstack().round(2)
```

```python
#find average ride length per member type per year 
year_data.groupby(['member_type', 'ride_year'],observed=True)['ride_length_minutes'].mean().unstack().round(2)
```

TIME OF DAY ANALYSIS

```python
#create a filter for morning between 5AM-11AM and get ride count per user type at this time 
year_data['started_at'] = pd.to_datetime(year_data['started_at'])

morning=(year_data.started_at.dt.hour>=5) & (year_data.started_at.dt.hour<=11)
morning=year_data[morning]
morning['member_type'].value_counts()
```

```python
#create a filter for afternoon between 11AM-5PM and get ride count per user type at this time 

year_data['started_at'] = pd.to_datetime(year_data['started_at'])

afternoon=(year_data.started_at.dt.hour>11) & (year_data.started_at.dt.hour<=16)
afternoon=year_data[afternoon]
afternoon['member_type'].value_counts()
```

```python
#create a filter for evening between 5pm-9pm and get ride count per user type at this time 

evening=(year_data.started_at.dt.hour>16) & (year_data.started_at.dt.hour<=21)
evening=year_data[evening]
evening['member_type'].value_counts()
```

```python
#create a filter for night between 9PM and 5AM and get ride count per user type at this time 

night=(year_data.started_at.dt.hour>21) & (year_data.started_at.dt.hour<24) | (year_data.started_at.dt.hour<5)
night=year_data[night]
night['member_type'].value_counts()
```
